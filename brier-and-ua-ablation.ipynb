```python
import os, subprocess
REPO = "https://github.com/Umer-3085/COMPASS.git"
DIR = "/kaggle/working/COMPASS"
if not os.path.exists(DIR):
    subprocess.run(["git", "clone", "--depth", "1", REPO, DIR], check=True)
else:
    subprocess.run(["git", "pull"], cwd=DIR, check=True)
os.chdir(DIR)

subprocess.run(["python", "-m", "src.kaggle_setup"], check=True)
DATA = open("/kaggle/working/compass_data/_data_root.txt").read().strip()
print("DATA =", DATA)
```

```python
import glob

# Auto-find the uploaded ckpt_best.pt anywhere under /kaggle/input/
ckpt_candidates = glob.glob("/kaggle/input/**/ckpt_best.pt", recursive=True)
assert len(ckpt_candidates) > 0, "Could not find ckpt_best.pt under /kaggle/input! Did the upload finish?"

s2_best = ckpt_candidates[0]
print("Found checkpoint at:", s2_best)
```

```python
subprocess.run([
    "python", "-m", "src.evaluate",
    "--ckpt", s2_best,
    "--data-root", DATA,
    "--out", "/kaggle/working/runs/eval_fixed",
    "--batch-size", "8", "--num-workers", "2"
], check=True)
```

```python
subprocess.run([
    "python", "-m", "src.evaluate",
    "--ckpt", s2_best,
    "--data-root", DATA,
    "--out", "/kaggle/working/runs/eval_ua",
    "--batch-size", "8", "--num-workers", "2",
    "--no-ug-gate"
], check=True)
```
